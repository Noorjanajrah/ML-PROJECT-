```python
from pathlib import Path
import kagglehub
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score, davies_bouldin_score

# 1. Download dataset via kagglehub and load CSV
path = kagglehub.dataset_download("saurabhbadole/wholesale-customers-data")
csv_files = list(Path(path).glob("*.csv"))
if not csv_files:
    raise FileNotFoundError(f"No CSV file found inside {path}")

df = pd.read_csv(csv_files[0])
print("Loaded:", csv_files[0], df.shape)
print("\nColumns and missing values:\n", df.isna().sum())

spend_cols = ["Fresh", "Milk", "Grocery", "Frozen", "Detergents_Paper", "Delicassen"]
print("\nSpending summary (distributions & outliers):\n", df[spend_cols].describe().round(1))

# 2. log1p then StandardScaler (fit on transformed spending features)
# log1p reduces right-skewness and handles zero values safely; StandardScaler equalizes feature scales
X_log = np.log1p(df[spend_cols])
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_log)

# 3 & 4. Run 8 experiments (4 KMeans + 4 Agglomerative Ward) and evaluate
exp_rows = []
fitted_labels = {}

for k in [2, 3, 4, 5]:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels_km = km.fit_predict(X_scaled)
    name_km = f"KMeans (k={k})"
    fitted_labels[name_km] = labels_km
    exp_rows.append({
        "Model": name_km,
        "Clusters": k,
        "Silhouette": silhouette_score(X_scaled, labels_km),
        "Davies-Bouldin": davies_bouldin_score(X_scaled, labels_km),
    })

for k in [2, 3, 4, 5]:
    agg = AgglomerativeClustering(n_clusters=k, linkage="ward")
    labels_agg = agg.fit_predict(X_scaled)
    name_agg = f"Agglomerative Ward (k={k})"
    fitted_labels[name_agg] = labels_agg
    exp_rows.append({
        "Model": name_agg,
        "Clusters": k,
        "Silhouette": silhouette_score(X_scaled, labels_agg),
        "Davies-Bouldin": davies_bouldin_score(X_scaled, labels_agg),
    })

exp_df = pd.DataFrame(exp_rows)
print("\nClustering Experiment Comparison:\n", exp_df.round(4).to_string(index=False))

# 5. Select highest Silhouette; tie-breaker: lowest Davies-Bouldin
exp_sorted = exp_df.sort_values(
    by=["Silhouette", "Davies-Bouldin"], ascending=[False, True]
).reset_index(drop=True)
best_exp_name = exp_sorted.loc[0, "Model"]
best_labels = fitted_labels[best_exp_name]
print("\nSelected clustering experiment:", best_exp_name)

# 6 & 7. Cluster sizes and original-scale mean/median spending per cluster
df_seg = df.copy()
df_seg["Cluster"] = best_labels

print("\nCluster sizes:\n", df_seg["Cluster"].value_counts().sort_index())
print("\nMean spending per cluster (original scale):\n", df_seg.groupby("Cluster")[spend_cols].mean().round(0))
print("\nMedian spending per cluster (original scale):\n", df_seg.groupby("Cluster")[spend_cols].median().round(0))

# Optional post-clustering inspection of Channel and Region (excluded from clustering)
for extra_col in ["Channel", "Region"]:
    if extra_col in df_seg.columns:
        print(f"\n{extra_col} distribution by Cluster:\n", pd.crosstab(df_seg["Cluster"], df_seg[extra_col]))

# 10. PCA 2D visualization (computed on X_scaled before plotting)
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)
print(f"\nExplained variance ratio (PC1 + PC2): {pca.explained_variance_ratio_.sum():.4f}")

plt.figure(figsize=(7, 5))
for c in sorted(np.unique(best_labels)):
    mask = best_labels == c
    plt.scatter(
        X_pca[mask, 0],
        X_pca[mask, 1],
        label=f"Cluster {c} (n={mask.sum()})",
        alpha=0.7,
        s=25,
    )
plt.xlabel(f"PC 1 ({pca.explained_variance_ratio_[0]*100:.1f}%)")
plt.ylabel(f"PC 2 ({pca.explained_variance_ratio_[1]*100:.1f}%)")
plt.title(f"PCA Visualization — {best_exp_name}")
plt.legend()
plt.tight_layout()
plt.show()
```
